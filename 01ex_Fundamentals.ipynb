2\. Convert the following function into a pure function with no global variables or side effects

````python
x = 5
def f(alist):
    for i in range(x):
         alist.append(i)
    return alist

alist = [1,2,3]
ans = f(alist)
print (ans)
print (alist) # alist has been changed!
````

````python
def f(alist, x):
    import copy
    lista=copy.copy(alist)
    for i in range(x):
        lista.append(i)
    return lista
    
alist = [1,2,3]
ans = f(alist, 5)
print (ans)
print (alist)
````
