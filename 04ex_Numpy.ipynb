1\. Find the row, column and overall means for the following matrix:

```python
m = np.arange(12).reshape((3,4))
```

````python
import numpy as np
m = np.arange(12).reshape((3,4))
print(m, '\n')

for i in range(3):
    print('mean for raw',i,':', np.mean(m[i,:]))
for j in range(4):
    print('mean for column',j, ':', np.mean((m[:,j])))
    
````

2\. Find the outer product of the following two vecotrs

```python
u = np.array([1,3,5,7])
v = np.array([2,4,6,8])
```

Do this in the following ways:

   * Using the function outer in numpy
   * Using a nested for loop or list comprehension
   * Using numpy broadcasting operatoins


````python
u = np.array([1,3,5,7])
v = np.array([2,4,6,8])

#outer
uv=np.outer(u,v)
print(uv, '\n')

#loop
l=[]
for i in range(len(u)):
    row=[]
    for j in range(len(v)):
        row.append(v[j]*u[i])
    l.append(row)
uv=np.array(l)
print(uv, '\n')
#broadcast
uv=np.tile(np.arange(1, 8, 2), (4, 1)).T
print( uv * v, '\n')
````

3\. Create a 10 by 6 matrix of random uniform numbers. Set all rows with any entry less than 0.1 to be zero

Hint: Use the following numpy functions - np.random.random, np.any as well as Boolean indexing and the axis argument.

````python
rows=[]
for i in range(10):
    row=[]
    for j in range(6):
        x=np.random.random()
        row.append(x)
    rows.append(row)
m=np.array(rows)
print('\n',m)

for i in range(len(m)):
    mask=(m[i,:]<0.1)
    if np.any(mask):
        m[i,:]=0
print(m, '\n')
````

4\. Use np.linspace to create an array of 100 numbers between 0 and 2π (includsive).

  * Extract every 10th element using slice notation
  * Reverse the array using slice notation
  * Extract elements where the absolute difference between the sine and cosine functions evaluated at that element is less than 0.1
  * Make a plot showing the sin and cos functions and indicate where they are close

````python
#import numpy.random as npr
a=np.linspace(0,2*np.pi,100)
#npr.shuffle(a)
print(a,'\n')

#Extract every 10th element using slice notation
for i in range(10):
    ten_elements=a[10*i: 10*i+10]
    print(ten_elements, '\n')
    
#Reverse the array using slice notation
a_reverse=a[::-1]
print(a_reverse,'\n')

#Extract elements where the absolute difference between the sine and cosine functions evaluated at that element is less than 0.1
print('elements where the absolute difference between the sine and cosine functionsis less than 0.1:',[element for element in a if np.abs(np.sin(element)-np.cos(element))<0.1])

#Make a plot showing the sin and cos functions and indicate where they are close
    #sin=np.array([np.sin(element) for element in a])
    #cos=np.array([np.cos(element) for element in a])
sin=np.sin(a)
cos=np.cos(a)
mask=(np.abs(sin-cos)<0.1)
equality_values=a[mask]

print('for the values', equality_values, 'the sine and cosine functions are equal')

import matplotlib.pyplot as plt
plt.plot(a, sin, 'r', label='Sine')
plt.plot(a, cos, 'b', label='Cosine')
plt.legend()
plt.xlabel('a')
plt.title('Sine(a) versus Cosine(a)')
plt.show()

````

5\. Create a matrix that shows the 10 by 10 multiplication table.

 * Find the trace of the matrix
 * Extract the anto-diagonal (this should be ```array([10, 18, 24, 28, 30, 30, 28, 24, 18, 10])```)
 * Extract the diagnoal offset by 1 upwards (this should be ```array([ 2,  6, 12, 20, 30, 42, 56, 72, 90])```)

````python
a=np.arange(1,11)
m=np.outer(a,a)
print(m, '\n')
anti_diagonal=[]
downward_diagonal=[]
for i in range(len(m)):
    raw=m[i,:]
    element=raw[-(i+1)]
    anti_diagonal.append(element)
for i in range(1,len(m)):
    raw=m[i,:]
    element=raw[i-1]
    downward_diagonal.append(element)
    
a_diag=np.array(anti_diagonal)
down_diagonal=np.array(downward_diagonal)
print(a_diag)
print(down_diagonal)


    
````

6\. Use broadcasting to create a grid of distances

Route 66 crosses the following cities in the US: Chicago, Springfield, Saint-Louis, Tulsa, Oklahoma City, Amarillo, Santa Fe, Albuquerque, Flagstaff, Los Angeles
