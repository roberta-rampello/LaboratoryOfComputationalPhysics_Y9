1\. Write a function that converts number representation, bin<->dec<->hex. (Clearly using the corresponding python built-in functions is not fair..)

```python
def dec_bin(x):
    if x == 0:
        return '0'
    lista = []
    while x > 0:
        d= x%2
        x=x//2
        lista.append(d)
    lista.reverse()
    bin= ''
    for element in lista:
        bin+='%s' % element
    return bin

def bin_dec(bin):
    i=0
    dec=0
    while bin > 0:
        dec += (bin % 10) * 2**i
        i += 1
        bin = bin // 10
    return dec

def dec_hex(x):
    if x == 0:
        return '0'
    lista = []
    while x > 0:
        d = x % 16
        if d == 10: d = 'A'
        elif d == 11: d = 'B'
        elif d == 12: d = 'C'
        elif d == 13: d = 'D'
        elif d == 14: d = 'E'
        elif d == 15: d = 'F'
        x = x // 16
        lista.append(d)
    lista.reverse()
    hex = ''
    for element in lista:
        hex += '%s' % element
    return hex
      
def hex_dec(hex):
    dec=0 
    lista=[]
    for character in hex: lista.append(character)
    lista.reverse() 
    for i in range(len(lista)): 
        char = lista[i]
        if char == 'A': char = 10
        elif char == 'B': char = 11
        elif char == 'C': char = 12
        elif char == 'D': char = 13
        elif char == 'E': char = 14
        elif char == 'F': char = 15
        else: char = int(char)
        dec += char * 16**i
    return dec

print(dec_bin(50))
print(bin_dec(110010))

print(dec_hex(16))
print(hex_dec('10'))

```

2\. Write a function that converts a 32 bit word into a single precision floating point (i.e. interprets the various bits as sign, mantissa and exponent)

```python
def single_precision_converter(bin_word):
    lista=[]
    for character in bin_word: 
        lista.append(int(character))
    sign=lista[0]
    exponent=lista[1:9] 
    mantissa=lista[9:]
    number=1
    k=-1
    for i in range(len(mantissa)):
        number+=mantissa[i]*2**k
        k+=-1
    real_exp=-127
    exponent.reverse()
    for i in range (len(exponent)) : 
        real_exp+=exponent[i]*2**i
    single_prec=((-1)**sign) *number*2**real_exp
    
    return single_prec

print(single_precision_converter('11000000101100000000000000000000'))
print(single_precision_converter('01000001101011110000000000000000'))
```

4\. Write a program to determine the machine precision

**Tips**: define a new variable by adding a smaller and smaller value (proceeding similarly to prob. 2) to an original variable and check the point where the two are the same 

```python
x=1
precision=1
last_precision=precision
while x + precision != 1:
    last_precision=precision
    precision=precision/2
    
print('the machine precision is', last_precision)
    
```

5\. Write a function that takes in input three parameters $a$, $b$ and $c$ and prints out the two solutions to the quadratic equation $ax^2+bx+c=0$ using the standard formula:
$$
x=\frac{-b\pm\sqrt{b^2-4ac}}{2a}
$$

(a) use the program to compute the solution for $a=0.001$, $b=1000$ and $c=0.001$

(b) re-express the standard solution formula by multiplying top and bottom by $-b\mp\sqrt{b^2-4ac}$ and again find the solution for $a=0.001$, $b=1000$ and $c=0.001$. How does it compare with what previously obtained? Why?

(c) write a function that compute the roots of a quadratic equation accurately in all cases

```python
def quadratic_equation(a,b,c):
    import math
    delta=math.sqrt(b**2-4*a*c) #if a*c << b, we have delta≈b, therefore  
    x1=(-b+delta)/(2*a)         # x1≈0 = -b+b, but...
    x2=(-b-delta)/(2*a)         # x2=-2b 

    x1_new= (2*c)/(-b-delta)    #here happens the contrary
    x2_new= (2*c)/(-b+delta)
    return x1, x2, x1_new, x2_new
    
x1, x2, x1_new, x2_new=quadratic_equation(0.001, 1000, 0.001)
print(x1, x2, x1_new, x2_new)
print('Is x1= x1_new?', x1==x1_new)
print('Is x2= x2_new?', x2==x2_new)

def new_quadratic_equation(a,b,c):
    import math
    delta=math.sqrt(b**2-4*a*c)  

    if b>=0: 
        x1=(2*c)/(-b-delta)  
        x2=(-b-delta)/(2*a) 
    else:
        x1=(-b+delta)/(2*a)   
        x2= (2*c)/(-b+delta)
        
    return x1, x2
x1, x2= new_quadratic_equation(0.001, 1000, 0.001)
print(x1, x2)
```

6\. Write a program that implements the function $f(x)=x(x−1)$

(a) Calculate the derivative of the function at the point $x = 1$ using the derivative definition:

$$
\frac{{\rm d}f}{{\rm d}x} = \lim_{\delta\to0} \frac{f(x+\delta)-f(x)}{\delta}
$$

with $\delta = 10^{−2}$. Calculate the true value of the same derivative analytically and compare with the answer your program gives. The two will not agree perfectly. Why not?

(b) Repeat the calculation for $\delta = 10^{−4}, 10^{−6}, 10^{−8}, 10^{−10}, 10^{−12}$ and $10^{−14}$. How does the accuracy scales with $\delta$?

```python
def f(x):
    y=x*(x-1)
    return y
    
def der_f(x, delta):
    der_y= (f(x+delta) - f(x))/delta

    analytic_der_y= 2*x -1
    return der_y, analytic_der_y

der_y, analytic_der_y= der_f(1, 1e-2)

print('The derivative via limit is=', der_y, '\nThe analytic derivative is =', analytic_der_y, '\nAre the two derivatives the same?', der_y==analytic_der_y)
lista=[1e-2, 1e-4, 1e-6, 1e-8, 1e-10, 1e-12, 1e-14]
for delta in lista: 
    der_y, analytic_der_y= der_f(1, delta)
    print('\nThe derivative via limit (delta=%s) is=' %delta, der_y, '\nThe analytic derivative is =', analytic_der_y, '\nAre the two derivatives the same?', der_y==analytic_der_y)
```

7\. Consider the integral of the semicircle of radius 1:
$$
I=\int_{-1}^{1} \sqrt(1-x^2) {\rm d}x
$$
which it's known to be $I=\frac{\pi}{2}=1.57079632679...$.
Alternatively we can use the Riemann definition of the integral:
$$
I=\lim_{N\to\infty} \sum_{k=1}^{N} h y_k 
$$

with $h=2/N$ the width of each of the $N$ slices the domain is divided into and where
$y_k$ is the value of the function at the $k-$th slice.

(a) Write a programe to compute the integral with $N=100$. How does the result compares to the true value?

(b) How much can $N$ be increased if the computation needs to be run in less than a second? What is the gain in running it for 1 minute? 


```python
def f(x):
    import math
    y=math.sqrt(1-x**2)
    return y
    
def Riemann_integral(N):
    h=2/N
    sum=0
    for k in range(1, N+1):
        x=-1+k*h
        sum+=h*f(x)
        
    return sum

print(Riemann_integral(100))

'''
import time

N_init = 100
N = N_init 

start = time.time()
Riemann_integral(N)
end = time.time()
execution_time = end - start

while execution_time < 1:
  N = N + 100000 
  start = time.time()
  Riemann_integral(N)
  end = time.time()
  execution_time = end - start
  print('Trying N =', N , 'the execution time is:',execution_time,'seconds')
print('With N=',N ,'the execution takes more than 1 sec' )'''
# I ran this with an increase of 100000 to get the interval of N in which i approah 1 second: now i filter a little better: Trying N = 2800100 the execution time is: 0.9843850135803223 seconds Trying N = 2900100 the execution time is: 1.0566961765289307 seconds With N= 2900100 the execution takes more than 1 sec

import time

N_init = 2800000
N = N_init 

start = time.time()
Riemann_integral(N)
end = time.time()
execution_time = end - start

while execution_time < 1:
  N = N + 1000 
  start = time.time()
  Riemann_integral(N)
  end = time.time()
  execution_time = end - start
  print('Trying N =', N , 'the execution time is:',execution_time,'seconds')
print('With N=',N ,'the execution takes more than 1 sec' )

print(Riemann_integral(N-1000))
```
