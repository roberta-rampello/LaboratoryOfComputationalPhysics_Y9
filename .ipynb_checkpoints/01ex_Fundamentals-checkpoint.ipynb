2\. Convert the following function into a pure function with no global variables or side effects

````python
x = 5
def f(alist):
    for i in range(x):
         alist.append(i)
    return alist

alist = [1,2,3]
ans = f(alist)
print (ans)
print (alist) # alist has been changed!
````

````python
def f(alist, x):
    import copy
    lista=copy.copy(alist)
    for i in range(x):
        lista.append(i)
    return lista
    
alist = [1,2,3]
ans = f(alist, 5)
print (ans)
print (alist)
````

3\. Write a `decorator` hello that makes every wrapped function print “Hello!”, i.e. something like:

```python
@hello
def square(x):
    return x*x
```



````python
def hello(func):
        def wrapper(*args, **kwargs):
            print('Hello')
            func(*args, **kwargs)
        return wrapper

@hello
def square(x):
    print(x**2)
    return 
@hello
def exponent(b, e):
    print(b**e)
    return
    
@hello    
def scritta():
    print('vediamo se ho capito')
    return
    
square(3)
exponent(2,4)
scritta()
````

4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
#a
def recursive_factorial(x):
    if x==0 or x==1:
        return 1
    else:
        return  x*recursive_factorial(x-1)
print(recursive_factorial(5))

#b
def factorial(x):
    prodotto=x
    for i in range(1,x):
        prodotto=prodotto*i
    return prodotto
        
print(factorial(5))
````

5\. Use HOFs (zip in particular) to compute the weight of a circle, a disk and a sphere, assuming different radii and different densities:

```python
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]
```

where the entries of the dictionary's values are the linear, superficial and volumetric densities of the materials respectively.

In particular define a list of three lambda functions using a comprehension that computes the circumference, the area and the volume for a given radius.


````python
import math
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]

type=(lambda r: 2*math.pi*r, lambda r: math.pi*r**2, lambda r: (4/3)*math.pi*r**3)
massa=('massa lineare', 'massa superficiale', 'massa volumetrica')
for metal, density in densities.items():
    print('\nMetal=%s'%(metal))
    for r in radii:
        print('Raggio= %d' %r)
        for d, kind, mas in zip(density, type, massa):
            m= d*kind(r)
            print(mas, ':',m)
````

6\. Edit the class defintion to add an instance attribute of is_hungry = True to the Dog class. Then add a method called eat() which changes the value of is_hungry to False when called. Figure out the best way to feed each dog and then output “My dogs are hungry.” if all are hungry or “My dogs are not hungry.” if all are not hungry. The final output should look like this:

`I have 3 dogs. 
Tom is 6. 
Fletcher is 7. 
Larry is 9. 
And they're all mammals, of course. 
My dogs are not hungry.
`


````python

# Parent class
class Dog:

    # Class attribute
    species = 'mammals'

    # Initializer / Instance attributes
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.is_hungry = True

    def eat(self):
        self.is_hungry= False

    # instance method
    def description(self):
        return "{} is {} years old".format(self.name, self.age)

    # instance method
    def speak(self, sound):
        return "{} says {}".format(self.name, sound)

# Child class (inherits from Dog class)
class RussellTerrier(Dog):
    def run(self, speed):
        return "{} runs {}".format(self.name, speed)

# Child class (inherits from Dog class)
class Bulldog(Dog):
    def run(self, speed):
        return "{} runs {}".format(self.name, speed)
        
def count(*args, **kwargs):
    lista=list(args) + list(kwargs.values())
    number_of_dogs=len(lista)
    return 'I have %s dogs' %number_of_dogs 

def status(*args, **kwargs):
    lista=list(args) + list(kwargs.values())
    hungry=all(dog.is_hungry==True for dog in lista)
    not_hungry=all(dog.is_hungry==False for dog in lista)
    if hungry:
        status='My dogs are hungry.' 
    if not_hungry:
        status='My dogs are not hungry.'
    return status
    
def feed(*args, **kwargs):
    lista=list(args) + list(kwargs.values())
    for dog in lista:
        dog.eat()
    return 

###
    
dog1=Dog('Tom', 6)
dog2=Dog('Fletcher', 7)
dog3=Dog('Larry', 9)

hungry= status(dog1, dog2, dog3)


number=count(dog1, dog2, dog3)
print(number)
print(dog1.description())
print(dog2.description())
print(dog3.description())

print("They're all %s, of course." %Dog.species) 

print(hungry)

feed(dog1, dog2, dog3)

hungry2= status(dog1, dog2, dog3)
print(hungry2)
````
