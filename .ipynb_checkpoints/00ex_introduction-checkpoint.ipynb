1. The MickeyMouse problem

a) Write a program that prints the numbers from 1 to 100. But for multiples of 3 print `Mickey` instead of the corresponding number and for the multiples of 5 print `Mouse`. For numbers which are multiples of both three and five print `MickeyMouse`

b) Put the result in a tuple and substitute `Mickey` with `Donald` and `Mouse` with `Duck`

```python
#(a)
lista=[]
for i in range(1, 101):
    if i%3==0 and i%5==0: 
        print('\n MickeyMouse')
        k='MickeyMouse'
    elif i%3==0:
        print('\n Mickey')
        k='Mickey'
    elif i%5==0:
        print('\n Mouse')
        k='Mouse'
    else:
        print('\n', i)
        k=i
    lista.append(k)
#(b)
for i in range(len(lista)):
    if lista[i]=='Mickey':
        lista[i]='Donald'
    elif lista[i]=='Mouse':
        lista[i]='Duck'   
    elif lista[i]=='MickeyMouse':
        lista[i]='DonaldDuck'
tupla=(lista)
print(tupla)
```

2\. The swap function

Write a function that swap the values of two input variables x and y (whatever the type). Try to do that also without a temporary variable

```python
def swap_function(x,y):
    t=y
    s=x
    new_x=t
    new_y=s
    return (new_x, new_y)

x=input('insert a value: x=')
y=input('insert a value: y=')
new_x, new_y=swap_function(x,y)
print('Values have been swapped: x=%s, y=%s' %(new_x, new_y))
    
```

3\. Computing the distance

Write a function that calculates and returns the euclidean distance between two points *u* and *v*, where *u* and *v* are both 2-tuples *(x,y)*. For example, if *u=(3,0)* and *v=(0,4)*, the function should return 5

```python
#u=(ux, uy)
#v=(vx, vy)
def Euclidean_distance(u, v): #u=(ux, uy) and v=(vx, vy)
    delta_x=u[0]-v[0]
    delta_y=u[1]-v[1]
    distance=(delta_x**2 +delta_y**2)**(1/2)
    return(distance)
distanza=Euclidean_distance((3,0),(0,4))
print(distanza)
```

4\. Counting letters

Write a program to calculate the number of times each character occurs in a given string *s*. Ignore differneces in capitalization

```python
def character_calculator(s):
    characters = [x for x in s.lower()]
    new_characters=[]
    counts=[]
    for i in range(len(characters)):
        if characters[i] not in new_characters and characters[i]!=' ':
            conteggi=characters.count(characters[i])
            new_characters.append(characters[i])
            counts.append(conteggi)
    dizionario=dict(zip(new_characters, counts))
   
    return dizionario
print(character_calculator('Scarpe piene di SassI'))
    
            


        
```

6\. Combination of functions

Write two functions - one that returns the square of a number, and one that returns the cube. Now write a third function that returns the number raised to the 6th power using the two previous functions.

```python
def square(n):
    return(n**2)
    
def cube(n):
    return(n**3)

def sixth_power(n):
    return(square(cube(n)))
print(sixth_power(2))
           
           
```

8\. Nested list comprehension

A Pythagorean triple is an integer solution to the Pythagorean theorem $a^2+b^2=c^2$. The first Pythagorean triple is (3,4,5). Find and put in a tuple all unique Pythagorean triples for the positive integers a, b and c less than 100.

```python
lista = []
for c in range(1, 101):
    for b in range(1, c):
        for a in range(1, b):
            if a**2 + b**2 == c**2:
                lista.append((a, b, c))
print(lista)
```

9\. Normalization

Write a function that takes a tuple of numbers and returns it with the entries normalized to one

```python
def normalization(tupla):
    somma=0
    t=[]
    for element in tupla:
        somma+=element
    if somma==0:
        errore=print('the sum of numbers is zero: impossible to normalize the tuple')
        return errore
    for k in range(len(tupla)):
        new_element=tupla[k]/somma
        t.append(new_element)
    tup=tuple(t)
    return (tup)

prova=normalization((10,30,20,25,15))
print(prova)
            
            
        
```
